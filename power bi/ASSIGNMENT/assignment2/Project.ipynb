```python
#1)using pie chart for diff test results

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df=pd.read_csv(r"C:\Users\Pugazharasan\Downloads\archive\healthcare_dataset.csv")
Normal_df=df[df["Test Results"]=="Normal"]
x=np.array(Normal_df["Test Results"])
Abnormal_df=df[df["Test Results"]=="Abnormal"]
y=np.array(Abnormal_df["Test Results"])
Inconclusive_df=df[df["Test Results"]=="Inconclusive"]
z=np.array(Inconclusive_df["Test Results"])
X=[]

X.append(len(x))
X.append(len(y))
X.append(len(z))
print(X)

test_result=['Normal','Abnormal','Inconclusive']
col=['r','b','g']

plt.pie(X,
        labels=test_result,
        colors=col,
        startangle=90,
        shadow=True,
        explode=(0.5,0,0),
        autopct="%0.5f%%")
plt.title("Test Results of Patients")
plt.show()
```

```python
#2) useing pie chart for Admission types


Elective_df=df[df["Admission Type"]=="Elective"]
x=np.array(Elective_df["Admission Type"])
Urgent_df=df[df["Admission Type"]=="Urgent"]
y=np.array(Urgent_df["Admission Type"])
Emergency_df=df[df["Admission Type"]=="Emergency"]
z=np.array(Emergency_df["Admission Type"])
A=[]

A.append(len(x))
A.append(len(y))
A.append(len(z))
print(A)

Admission=['Elective','Emergency','Urgent']
col=['y','g','c']

plt.pie(A,
        labels=Admission,
        colors=col,
        startangle=90,
        shadow=True,
        explode=(0,0.1,0),
        autopct='%0f%%')
plt.title("Patient class")
plt.show()
```

```python
df=pd.read_csv(r"C:\Users\Pugazharasan\Downloads\archive\healthcare_dataset.csv")

fig, axes = plt.subplots(2, 2, figsize=(25, 16))
fig.suptitle('Healthcare Dataset Analysis', fontsize=11)
gender_counts = df['Gender'].value_counts()
gender_counts.plot(kind='bar', ax=axes[0, 0])
axes[0, 0].set_title('Count of Male & Female')
axes[0, 0].set_xlabel('Gender')
axes[0, 0].set_ylabel('Count')
axes[0, 0].tick_params(axis='x', rotation=0)
for i, v in enumerate(gender_counts):
    axes[0, 0].text(i, v, str(v), ha='center', va='bottom')

male_affected = df[df['Gender'] == 'Male']['Medical Condition'].value_counts()
male_affected.plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title('Male Patients Affected by Condition')
axes[0, 1].set_xlabel('Medical Condition')
axes[0, 1].set_ylabel('Count')
axes[0, 1].tick_params(axis='x', rotation=45)
for i, v in enumerate(male_affected):
    axes[0, 1].text(i, v, str(v), ha='center', va='bottom')


female_affected = df[df['Gender'] == 'Female']['Medical Condition'].value_counts()
female_affected.plot(kind='bar', ax=axes[1,0])
axes[1, 0].set_title('Female Patients Affected by Condition')
axes[1, 0].set_xlabel('Medical Condition')
axes[1, 0].set_ylabel('Count')
axes[1, 0].tick_params(axis='x', rotation=45)
for i, v in enumerate(female_affected):
    axes[1, 0].text(i, v, str(v), ha='center', va='bottom')

bill_amount_by_gender = df.groupby('Gender')['Billing Amount'].sum()
bill_amount_by_gender.plot(kind='bar', ax=axes[1, 1])
axes[1, 1].set_title('Total Bill Amount by Gender')
axes[1, 1].set_xlabel('Gender')
axes[1, 1].set_ylabel('Total Bill Amount ($)')
axes[1, 1].tick_params(axis='x', rotation=45)
for i, v in enumerate(bill_amount_by_gender):
    axes[1, 1].text(i, v, f'${v:,}', ha='center', va='bottom')
```
